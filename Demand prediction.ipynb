# Step 1:Data Mining

```python
import pandas as pd
# Load the dataset
shirt_size = pd.read_csv("t-shirts.csv")
# Display basic information and statistics
print(shirt_size.info())
print(shirt_size.describe(include='all'))
print(shirt_size.head())
```



# Step 2: Data Preparation

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

# Encode categorical variables
label_encoder = LabelEncoder()
shirt_size['size'] = label_encoder.fit_transform(shirt_size['size'])
shirt_size['material'] = label_encoder.fit_transform(shirt_size['material'])
shirt_size['color'] = label_encoder.fit_transform(shirt_size['color'])
shirt_size['sleeves'] = label_encoder.fit_transform(shirt_size['sleeves'])
shirt_size['demand'] = label_encoder.fit_transform(shirt_size['demand'])

# Split the data
X = shirt_size.drop('demand', axis=1)
y = shirt_size['demand']
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, test_size=0.3, random_state=100)
```

# Data Processing Methods

```python
#Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
#Normalization
from sklearn.preprocessing import MinMaxScaler

normalizer = MinMaxScaler()
X_train_normalized = normalizer.fit_transform(X_train)
X_test_normalized = normalizer.transform(X_test)
```

#  Step 3: Classification

```python
from sklearn.naive_bayes import GaussianNB,MultinomialNB
from sklearn.tree import DecisionTreeClassifier

# Naive Bayes without scaling
nb = MultinomialNB(alpha=0.000001,force_alpha=True,fit_prior=True,)
nb.fit(X_train, y_train)


```

```python
y_predict = nb.predict(X_test)
nbtr_accracy=nb.score(X_train,y_train)*100
nbts_accuracy=nb.score(X_test,y_test)*100
print("Training Accuracy",nbtr_accracy)
print("Testing Accuracy",nbts_accuracy)
```

```python
# Decision Tree without scaling
dt = DecisionTreeClassifier(max_depth=100, random_state=200,ccp_alpha=0.00001,criterion='gini')
dt.fit(X_train, y_train)
```

```python
dttr_accuracy=dt.score(X_train,y_train)*100
dtts_accuracy = dt.score(X_test, y_test) * 100
print("Training Accuracy",dttr_accuracy)
print("Testing Accuracy",dtts_accuracy)
```

```python
# Naive bayes with standariztaion
X_train_scaled_=abs(X_train_scaled)
nbs=nb.fit(X_train_scaled_, y_train)
```

```python
nbstr_accuracy=nbs.score(X_train,y_train)*100
nbsts_accuracy = nbs.score(X_test, y_test) * 100
print("Training Accuracy",nbstr_accuracy)
print("Testing Accuracy",nbsts_accuracy)
```

```python
# Naive bayes with standariztaion

nbs=nb.fit(X_train_normalized, y_train)
```

```python
nbntr_accuracy=nbs.score(X_train,y_train)*100
nbnts_accuracy = nbs.score(X_test, y_test) * 100
print("Training Accuracy",nbntr_accuracy)
print("Testing Accuracy",nbnts_accuracy)
```

```python
dtstrain_accuracy=dts.score(X_train_scaled,y_train)*100
dtstest_accuracy = dts.score(X_test_scaled, y_test) * 100

print("Traing Accuracy= ",dtstrain_accuracy)
print("Testing Accuracy= ",dtstest_accuracy)
```

#  Bonus: Handle Missing Data


```python
import numpy as np

# Introduce missing values
np.random.seed(100)
missing_indices = np.random.choice(X_train.index, size=int(0.05*len(X_train)), replace=False)
X_train.loc[missing_indices] = np.nan

# Handle missing data
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='mean')
X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Evaluate with imputed data
dt.fit(X_train_imputed, y_train)
y_pred_dt_imputed = dt.predict(X_test_imputed)
print("Decision Tree with Imputed Data Classification Report:")
print(classification_report(y_test, y_pred_dt_imputed))
```
